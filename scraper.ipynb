```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
```

```python
r = requests.get("https://dustloop.com/w/GGST")
```

```python
r.status_code
```

```python
soup = BeautifulSoup(r.text)
print(soup.prettify())
```

```python
raw_links = soup.find_all("a")
```

```python
# Filtering the links to only include those that link to individual fighters' pages
filtered_raw_links = raw_links[49:83]
for i in range(len(filtered_raw_links)):
    print(f"{i}: {filtered_raw_links[i]["href"]}")
```

```python
links = [link["href"] for link in filtered_raw_links]
print(len(links))
# There are no duplicates, so a set is unnecessary.
```

```python
final_links = ["https://dustloop.com" + link for link in links]
print(final_links)
```

```python
r = requests.get(final_links[0])
soup = BeautifulSoup(r.text)
data = soup.find_all("div", {"class": "frameDataGridRow colCount-7"})
```

```python
fiveK = data[3].contents
fiveK = [thing.text for thing in fiveK]
fiveK = fiveK[2:6]
fiveK
```

```python
data_dict = {
    "name": [],
    "5P": []
}

def safe_scrape(soup, tag, class_name):
    try:
        return soup.find(tag, {"class": class_name}).text.strip()
    except:
        return None

def safe_scrape_moves(soup):
    try:
        soup.find_all("div", {"class": "frameDataGridRow colCount-7"})
    except:
        return None

def extract_move_data(moves, index):
    """
    0 = 5P   1 = 5K   2 = c.S   3 = f.S   4 = 5H   5 = 5D
    6 = 2P   7 = 2K   8 = 2S    9 = 2H    10 = 2D  11 = 6P
    """
    move = data[index].contents
    move = [attribute.text for attribute in move]
    relevant_info = move[1:6]
    return relevant_info

for link in final_links:
    print(link)
    directory = requests.get(link)
    soup = BeautifulSoup(directory.text)
    moves = safe_scrape_moves(soup)

    name = safe_scrape(soup, "div", "t-infobox-title")
    fiveP = extract_move_data(moves, 0)
    fiveK = extract_move_data(moves, 1)
    cS = extract_move_data(moves, 2)
    fS = extract_move_data(moves, 3)
    fiveH = extract_move_data(moves, 4)
    
    data_dict['name'].append(name)

    df = pd.DataFrame(data_dict)
df.head()
```
